```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "data/augmented_dataset.csv",
    parse_dates=["timestamp"],
)
# df.drop(columns=["temperature_raw", "growth_raw", "humidity_raw", "co2_raw"], inplace=True)
df.rename(columns={
    "temperature_norm": "temperature",
    "growth_norm": "growth",
    "humidity_norm": "humidity",
    "co2_norm": "co2", 
    "growth_rate_norm": "growth_rate",
    "acceleration_norm": "acceleration"
    }, inplace=True)

df.to_csv("data/processed/augmented_dataset.csv", index=False)
df.tail()
```

```python
subset = df[df["variation_id"] == 12].reset_index(drop=True)

plt.figure()
plt.plot(subset["growth"])

colors = {0: "blue", 1: "green", 2: "orange", 3: "red"}

for i in range(len(subset) - 1):
    stage = subset["stage"].iloc[i]
    plt.axvspan(i, i + 1, color=colors[stage], alpha=0.1)

plt.title("Growth with Stage Background")
plt.show()
```
